```python
def get_atr_place (driver):

    metro_stations = []

    try:
        price_element = driver.find_element(By.CSS_SELECTOR, 'span[itemprop="price"]')
        price = price_element.get_attribute("content")
    except:
        price = ""

    try:
        map_elem = driver.find_element(By.CSS_SELECTOR, "[class^='style-item-map-wrapper']")
        latitude = map_elem.get_attribute('data-map-lat')
        longitude = map_elem.get_attribute("data-map-lon")
    except:
        latitude = ""
        longitude = ""

    try:
        metro_stations_elem = driver.find_elements(By.CLASS_NAME, 'style-item-address-georeferences-item-TZsrp')

        for metro_station_elem in metro_stations_elem:

            match = re.match(r"([^\d]+)([\d–]+)", metro_station_elem.text) # регуляторку запросила у чата гпт "при помощи библиотеки re разбей строку текст на 2 части пусть все что до цифр будет в одном элементе строки, а все цифры и тире между ними в другой"
            station_name = match.group(1).strip()
            distance = match.group(2).strip()
            metro_stations.append([station_name, distance])
        
        #В конце надо будет руками проверить на "Улица 1905 года"
    except:
        metro_stations = []

    try:
        about_elems = driver.find_elements(By.CLASS_NAME, "params-paramsList__item-_2Y2O")
        for about_elem in about_elems:
            if "Общая площадь" in about_elem.text:
                try:
                    match = re.search(r'([\d.]+)', about_elem.text)
                    area = match.group(1)
                except:
                    area = ""
            if "Этаж" in about_elem.text:
                try:
                    match = re.search(r'([\d.]+)', about_elem.text)
                    floor = match.group(1)
                except:
                    floor = ""
            if "Минимальный срок аренды" in about_elem.text:
                try:
                    min_book = re.search(r'\d+\.?\d*\s*.*', about_elem.text).group() #опять гпт с регуляторкой помог
                except:
                    min_book = ""   
    except:
        area = ""
        floor = ""
        min_book = ""

    return price, latitude, longitude, metro_stations, area, floor, min_book
```

```python
service = Service()
options = webdriver.ChromeOptions()
driver = webdriver.Chrome(options=options, service = service)

prices = []
latitudes = []
longitudes = []
metro_stations = []
areas = []
floors = []
min_books = []


driver.get("https://www.avito.ru/moskva/kommercheskaya_nedvizhimost/sdam-ASgBAgICAUSwCNRW?f=ASgBAQECAkSwCNRW9BKk2gEBQJ7DDSSG2TnQ1Y4DAUW2ExR7ImZyb20iOjUwLCJ0byI6MTAwfQ")

if "Not Found" in driver.title:
  print("Can't found link")
else:
  for _ in range(10):
    places = driver.find_elements(By.CSS_SELECTOR, "[class^='iva-item-sliderLink-Fvfau']")
    for place in places:
      try:
        #https://stackoverflow.com/questions/66173720/elementclickinterceptedexception-message-element-click-intercepted-using-selen
        #https://stackoverflow.com/questions/41744368/scrolling-to-element-using-webdriver
        driver.execute_script("arguments[0].scrollIntoView(true);", place)
        time.sleep(2)

        WebDriverWait(driver, 10).until(EC.element_to_be_clickable(place))
        place.click()
        time.sleep(4)

        try:
          driver.switch_to.window(driver.window_handles[1])
          time.sleep(2)
          
          price, latitude, longitude, place_metro_stations, area, floor, min_book = get_atr_place (driver)
          
          prices.append(price)
          latitudes.append(latitude)
          longitudes.append(longitude)
          metro_stations.append(place_metro_stations)
          areas.append(area)
          floors.append(floor)
          min_books.append(min_book)

          driver.close()

          driver.switch_to.window(driver.window_handles[0])
          time.sleep(2)
        except:
          driver.switch_to.window(driver.window_handles[1])
          time.sleep(2)
          
          driver.close()

          driver.switch_to.window(driver.window_handles[0])
          time.sleep(2)

      except Exception as e:
        print(f"Ошибка при клике на элемент: {e}")

    button = WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "[class*='listItem_arrow_next']"))
    )
    driver.execute_script("arguments[0].scrollIntoView(true);", button)
    button.click()
```

```python
df = pd.DataFrame({
    'prices': prices,
    'latitudes': latitudes,
    'longitudes': longitudes,
    'metro_stations': metro_stations,
    'areas': areas,
    'floors': floors,
    'min_books': min_books
})

df.to_csv('C:/Users/User/Desktop/проект НоД/places.csv', index=False, encoding='cp1251')
```
